```julia
using Random
```

```julia
N = 50
data = rand(N, 4)
```

```julia
fields = hcat(
    mod.(data[:,1] + data[:,4], 1) .> mod.(data[:,2] + data[:,4], 1),
    mod.(data[:,1] + data[:,4], 1) .> data[:,3],
    mod.(data[:,1] + data[:,4], 1) .≤ mod.(data[:,2] + data[:,4], 1) + data[:,3],
    data[:,1] .> data[:,2],
    data[:,1] .> data[:,3],
    data[:,1] .≤ data[:,2] + data[:,3],
    data[:,1] .+ data[:,4] .≥ 1,
    data[:,2] .+ data[:,4] .≥ 1,
)
```

```julia
results = hcat(
    fields[:,1],
    fields[:,2],
    fields[:,7],
    fields[:,4] .& fields[:,5] .& fields[:,6],
    fields[:,1] .& fields[:,2] .& fields[:,3],
    fields[:,8],
    fields[:,4],
    fields[:,5],
)
```

```julia
sum(sum(results[:,1:4], dims = 2) .> sum(results[:,5:8], dims = 2))
```

```julia
data = floor.(rand(10, 3), digits = 1)
```

```julia
findall(
    (data[:,1] .> data[:,3]) 
    .& (data[:,2] .> data[:,3]) 
    .& ((data[:,1] + data[:,2]) .> data[:,3] .+ 1)
)
```
